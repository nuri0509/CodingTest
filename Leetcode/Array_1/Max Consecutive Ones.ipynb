(Example 1)
- Input: nums = [1,1,0,1,1,1]
- Output: 3
- Explanation: <br>
The first two digits or the last three digits are consecutive 1s. </br>
The maximum number of consecutive 1s is 3.<br>

- 1 <= nums.length <= 105

```python
# My code(기존 작성코드를 수정했음): 코드 스타일 유지
class Solution(object):
    def findMaxConsecutiveOnes(self, nums):
        cnt_list = []
        cnt = 0
        for i in nums:
            if i == 1:
                cnt += 1
            else:
                cnt_list.append(cnt)
                cnt = 0 # 다음 카운트를 위해 0으로 초기화해줘야 합니다!
                
        # [중요] for문이 끝나고 마지막으로 남아있던 cnt도 마저 넣어줍니다.
        cnt_list.append(cnt) 
                
        # 리스트 내장 메서드가 아닌 내장 함수 max() 사용
        return max(cnt_list)
```

```python
# Example - 실시간으로 최대값(max_cnt)를 갱신
class Solution(object):
    def findMaxConsecutiveOnes(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        max_cnt = 0  # 여태까지 본 가장 긴 연속된 1의 개수
        cnt = 0      # 현재 연속 중인 1의 개수
        
        for i in nums:
            if i == 1:
                cnt += 1
                # 1을 만날 때마다 실시간으로 최대값을 업데이트합니다.
                if cnt > max_cnt:
                    max_cnt = cnt
            else:
                # 0을 만나면 현재 카운트만 리셋합니다.
                cnt = 0
                
        return max_cnt
```
